```sql
SHOW TABLES IN banking_risk.bronze;

SELECT
    COUNT(*) AS total_transactions,
    COUNT(DISTINCT transaction_id) AS unique_transactions,
    MIN(transaction_timestamp) AS min_timestamp,
    MAX(transaction_timestamp) AS max_timestamp
FROM banking_risk.bronze.transactions;

SELECT
    transaction_id,
    _ingested_at
FROM banking_risk.bronze.transactions
LIMIT 10;
SELECT COUNT(*) FROM banking_risk.bronze.customers;

SELECT COUNT(*) FROM banking_risk.bronze.accounts;

SELECT COUNT(*) FROM banking_risk.bronze.merchants;


-- Silver Transactions

SELECT COUNT(*) AS silver_count
FROM banking_risk.silver.transactions;
SELECT COUNT(*) AS quarantine_count
FROM banking_risk.silver.transaction_quarantine;
SELECT
    transaction_id,
    COUNT(*) AS cnt
FROM banking_risk.silver.transactions
GROUP BY transaction_id
HAVING COUNT(*) > 1;
SELECT
    transaction_id,
    account_id,
    merchant_id,
    transaction_timestamp,
    amount,
    currency,
    country,
    status
FROM banking_risk.silver.transactions
ORDER BY transaction_timestamp
LIMIT 20;


-- Enrichment
SELECT *
FROM banking_risk.silver.enriched_transactions
LIMIT 20;
SELECT
    COUNT(*) AS total,
    COUNT(customer_id) AS with_customer,
    COUNT(merchant_id) AS with_merchant
FROM banking_risk.silver.enriched_transactions;



SELECT
    transaction_id,
    amount,
    high_value_flag,
    suspicious_merchant_flag,
    unusual_location_flag
FROM banking_risk.silver.risk_features
LIMIT 20;


SELECT *
FROM banking_risk.silver.velocity_features
ORDER BY transaction_count DESC
LIMIT 20;



SELECT
    transaction_id,
    amount,
    high_value_flag,
    velocity_flag,
    unusual_location_flag,
    suspicious_merchant_flag,
    risk_score,
    risk_classification
FROM banking_risk.gold.transaction_risk
ORDER BY risk_score DESC
LIMIT 20;
```
